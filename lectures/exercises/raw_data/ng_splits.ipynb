```python
from pathlib import Path
import re

hp = open('hp.txt',      encoding='utf-8').read()
tk = open('tolkien.txt', encoding='utf-8').read()
```

```python
def clean(t):
    t = t.replace('«', '"').replace('»', '"').replace('–', '-').replace('…', '...').replace('‘',"'").replace('’',"'").replace('”','"').replace('“','"')
    t = re.sub(r'[<>{|}¦°˜¬■�•\t]', '', t)   # drop junk
    return t

def split(text, ratio=0.9):
    n = int(ratio * len(text))
    return text[:n], text[n:]
```

```python
hp = clean(hp)
tk = clean(tk)
hp_train, hp_val = split(hp)
tk_train, tk_val = split(tk)

SEP = '\n\n\n'
train = hp_train + SEP + tk_train
val   = hp_val   + SEP + tk_val

Path('../data/train_ng.txt').write_text(train, encoding='utf-8')
Path('../data/val_ng.txt').write_text(val, encoding='utf-8')

print(f'train: {len(train):>9,} chars  ({len(hp_train):>9,} HP + {len(tk_train):>9,} Tolkien)')
print(f'val:   {len(val):>9,} chars  ({len(hp_val):>9,} HP + {len(tk_val):>9,} Tolkien)')
```
