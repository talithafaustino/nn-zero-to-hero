```python
bands = pd.read_csv('metal_bands_raw.csv')
names = bands['Name']
names.shape
```

```python
# names = names.dropna()
# names = names[names.str.strip() != ""]
```

```python
names = names.drop_duplicates()
names.shape
```

```python
names = names[names.str.match(r'^[a-zA-Z0-9 ]+$')]
names.shape
```

```python
names = names[~names.str.match(r'^[0-9 ]+$')]
names.shape
```

```python
# Split all names into words and flatten the list
all_words = " ".join(names).split()

# Get the unique words using a set
unique_words = set(all_words)

# Count the number of distinct words
num_distinct_words = len(unique_words)

print(num_distinct_words)
```
